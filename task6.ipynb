print numbers from 10 to 30 using while loop

```python
start=10
while start<=30:
  print(start)
  start+=1
```

```python
# odd numbers from 1 to 20
st=1
while st<=20:
  if st%2!=0:
    print(st)
  st+=1
```

```python
# odd numbers from 1 to 20
st=1
while st<=20:
  print(st)
  st+=2
```

```python
word="hello"
count=5
while count>=1:
  print(word)
  count-=1
```

```python
first=5
while first>=1:
  print(first)
  first-=1
```

```python
# all multiples of 3 upto 30 
number=3
start=1
while start <=10:
  print(f"{number*start}")
  start+=1
```

```python
# keeep asking for a number until it is a postive number 

keep_asking=True

while keep_asking:
  positive_number=int(input("ENter a positive number"))
  if positive_number>0:
    print("thank you for the postive number",positive_number)
    keep_asking=False

```

```python
# guess the secret number
secret=7
is_correct=True
while is_correct:
  guess=int(input("Guess"))
  if guess==secret:
    print("secret number is ",guess)
    is_correct=False
```

```python
# function to add two numbers
def add(a,b):
  return a+b

print(add(2,3))
```

```python
# lambda version
add_lambda=lambda a,b:a+b
print(add_lambda(2,3))
```

```python
# function to check odd or even

def isEven(num):
  if num%2==0:
    return True
  else:
    return False
  
print(isEven(7))
```
