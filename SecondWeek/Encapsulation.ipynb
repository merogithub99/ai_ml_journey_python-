Encapsulation is the process of binding the data inside the class or method managing the access to the user 
public means ----> name
protected -------> _name
private ---------> __name 

```python
class Bank:
  bank_name="NIMB"    #this is instance variabale

  def __init__(self,owner,balance):
    self.owner=owner
    self.__balance =balance


  def deposit(self,amount):
    if amount>0:
      self.__balance+=amount
      print(f"{amount} deposited succesfully!")
    else:
      print("amount must be more than zero")

  def withdraw(self,amount):
    if self.__balance>=amount:
      self.__balance-=amount
      print(f"{amount} withdrawan successfully")
    else:
      print("Insufficient fund ")

  def get_balance(self):
    return self.__balance
  

acc1=Bank("ram",10000)
acc1.get_balance()


```
