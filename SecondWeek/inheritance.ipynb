```python
class Animal:
  def __init__(self,name):
    self.name=name

  def sound(self):
    return f"{self.name} makes a sound"
```

```python
class Cat(Animal):
  def sound(self):
    # return super().sound()
    return f"{self.name} speak meow"
  
class Dog(Animal):
  def sound(self):
    return f"{self.name} speak howhow"
```

```python
c=Cat("cat1")
c.sound()
```

```python
d=Dog("dog1")
d.sound()
```
