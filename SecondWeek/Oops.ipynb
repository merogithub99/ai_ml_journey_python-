```python
class Car:
  def __init__(self,speed,color,has_break):
    self.speed=speed
    self.color=color
    self.has_break=has_break

  def car_attr(self):
    return self.__dict__

car1=Car(10,"red",True)
print(car1.has_break)
print(car1.speed)
print(car1.color)

  
```

return everything inside the object through dictionary

```python
print(Car.car_attr(car1))
```

magic method

```python
class Bus:
  def __init__(self,speed,color,has_break):
    self.speed=speed
    self.color=color
    self.has_break=has_break

  def bus_result(self):
   return {
     "speed": self.speed,
     "color":self.color,
    "has_break":self.has_break
      }
  
  def __str__(self):
   return f"speed: {self.speed} has_color: {self.has_break} color:{self.color}"
  
bus=Bus(100,"blue",True)
print(bus)

```

returning through dictionary

```python
bus2=Bus(200,"red",True)
print(bus)
print(bus2.bus_result())
```
