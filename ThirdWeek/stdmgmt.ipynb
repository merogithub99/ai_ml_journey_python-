```python
students=[]
def add_student():
  try:
    id=input('Enter  student id ')
    name=input('Enter  student name ')
    age=input('Enter  student age ')
    grade=input('Enter  student grade ')

    students.append({'id':id,'name':name,'age':age,'grade':grade})

    print("students added")

  except:
    print("Invalid input")
```

```python
def view_students():
  if not  students:
    print("no students found")
    return
  
  for s in students:
    print(s['id'],s['name'],s['age'])


def search_students():
  id =input("Enter id to search:")


  for s in students:
    if s['id']==id:
      print(s)
      return
    
    print("student not found")
    
    
```

```python
def delete_students():
  id =input("Enter id to delete")


  for s in students:
    if s['id']==id:
      students.remove(s)
      print("students deleted")
      return
    
    print("students not found")
```

```python
while True:
  print("\n student mgmt system")

  print("\n 1 add student")
  print("\n 2 view student")
  print("\n 3 search student")
  print("\n 4 delete student")
  print("\n 5 exit ")

  choice=input("Enter your choice")

  if choice=="1":
    add_student()

  elif choice=="2":
    view_students()

  elif choice=="3":
    search_students()

  elif choice=="4":
    delete_students()

  elif choice=="5":
    print("exiting........")
    break
  
  else:
    print("invalid choice")
```
