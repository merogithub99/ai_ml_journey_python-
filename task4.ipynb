count vowels

```python
vowel_count=input("Enter a string to count vowel in ")
count=0
for i in vowel_count:
  if i=='a' or i=='e' or i=='i' or i=='o' or i=='u':
    count+=1
    print(f"{i} is vowel")
print(count)
```
