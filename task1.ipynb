filter only int and float from the list and store them in another list:


```python
list1=["ram",1,2,3,4.4,5.5,"hello",[1,2,3]]
list2=[]
for i in list1:
  if type(i) ==int or type(i)==float:
    list2.append(i)

list2
```

```python
list1=["ram",1,2,3,4.4,5.5,"hello",[1,2,3,"str"]]
list2=[]
for i in list1:
  if type(i)==list:
    for j in i:
      if type(j)== int or type(j)==float:
        list2.append(j)

  else:
     if type(i)==int or type(i)==float:
      list2.append(i)

list2
```
