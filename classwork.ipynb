```python
# prime number check 
def check_prime(number):
  if number<=1:
    return False
  for i in range(2,number):
    if number%i==0:
      return False
  else:
    return True
     
number =int(input("Enter your number"))

print(f" the number {number} is prime  number?{check_prime(number)}")
```

```python
import math 
def is_prime(n):
  if n>1:
    for i in range(2,int(math.sqrt(n))+1):
      if n%i==0:
        return False
        break
    return True
  
  else:
    return False
  
print(is_prime(53))
  
  
```

```python
# write a func that takes the list and return the second largest numbers\

def returnsecondlargest(lst):
  lst.sort()
  return lst[-2]

print(returnsecondlargest([1,2,3,5,4,6,10,9]))
```

```python
import math 
def check_prime_optimized(number):
    if number <= 1: return False
    if number == 2: return True  # 2 is the only even prime
    if number % 2 == 0: return False # Skip all other even numbers
    
    # Check only odd numbers up to the square root
    limit = int(math.sqrt(number)) + 1
    for i in range(3, limit, 2):
        if number % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
print(f"Is {num} prime? {check_prime_optimized(num)}")
```

```python
def check_prime_simple(number):
    if number <= 1:
        return False
    # Checks every number up to the input
    for i in range(2, number):
        if number % i == 0:
            return False
    else:
        return True

num = int(input("Enter a number: "))
print(f"Is {num} prime? {check_prime_simple(num)}")
```
