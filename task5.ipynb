elements of list in uppercase

```python
list_of_word=["apple","ball","cat","dog"]
for item in list_of_word:
  print(item.upper())

print(list_of_word)
print(list_of_word[0])
```

permanently making all items uppercase

```python
for i in range(len(list_of_word)):
  list_of_word[i]=list_of_word[i].upper()

print(list_of_word)
```

numbers greater than 10 in a list

```python
numlist=[21,342,34,1,2,3,4,45,5,7]
count_of_g10=0
for i in numlist:
  if i>10:
    count_of_g10+=1
print(count_of_g10)
```
